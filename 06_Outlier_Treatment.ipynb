## **1. What is an Outlier?**



**Understand the Concept**

- An outlier is a data value that is very different from most other values.
- Outliers can be unusually high or unusually low.
- An outlier is not always an error; it can be a valid extreme observation.




**Demonstrate the Concept**

**Example:**

- Check unusually high or low Study_Hours values.

**AI/ML Usage:**

- Helps decide the appropriate treatment for extreme values.

```python
# Implement the Concept
# study hours
df["Study_Hours"].describe()
```

**Explanation**

- describe() gives summary statistics for Study_Hours.
- It shows the minimum, maximum, mean, and quartile values.

## **2. Outlier vs Error**



**Understand the Concept**

- An outlier is an unusual value, but it can be valid.
- An error is an incorrect or impossible value.





**Demonstrate the Concept**

**Example:**

- A study time of 14 hours may be valid, while -5 hours is an error.

**AI/ML Usage:**

- Helps choose the correct treatment for unusual observations.

```python
# Implement the Concept
# Outlier 
Outlier = df[(df["Study_Hours"] == 0) |  (df["Study_Hours"] > 12)]
Outlier[["Student_ID", "Study_Hours"]].head(3)
```

```python
# Error 
df[df["Study_Hours"] < 0]
```

## **3. IQR Method**



**Understand the Concept**

- IQR method identifies outliers using the first quartile and third quartile.
- IQR is calculated as the difference between Q3 and Q1.
- Values below the lower limit or above the upper limit are treated as potential outliers.



**Demonstrate the Concept**

**Example:**

- Use the IQR method to detect outliers in study hours.

**AI/ML Usage:**

- Helps identify extreme values before model training.

```python
# Implement the Concept
# quartiles
Q1 = df["Study_Hours"].quantile(0.25)
Q3 = df["Study_Hours"].quantile(0.75)

# IQR
IQR = Q3 - Q1

# limits
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# outliers
outliers = df[(df["Study_Hours"] < lower) | (df["Study_Hours"] > upper)]
print("Lower Limit:", lower)
print("Upper Limit:", upper)
print("Outliers:", len(outliers))
```

```python
outliers[["Student_ID", "Study_Hours"]].tail(3)
```

**Explanation**

- Q1 and Q3 define the middle 50% of the data.
- IQR is used to calculate the lower and upper limits.
- Values outside these limits are identified as potential outliers.

## **4. Z-Score Method**



**Understand the Concept**

- Z-score shows how far a value is from the mean in terms of standard deviations.
- A value with a very high or low Z-score can be considered a potential outlier.
- Commonly, values with an absolute Z-score greater than 3 are treated as outliers.


**Demonstrate the Concept**

**Example:**

- Use Z-score to find extreme study hour values.

**AI/ML Usage:**

- Helps detect extreme values before model training.

```python
# Implement the Concept
# Z-score
mean = df["Study_Hours"].mean()
std = df["Study_Hours"].std()
z_score = (df["Study_Hours"] - mean) / std

# outliers
outliers = df[z_score.abs() > 3]
print("Outliers:", len(outliers))
```

```python
outliers[["Student_ID", "Study_Hours"]].sample(3)
```

**Explanation**

- Mean and standard deviation are used to calculate the Z-score.
- abs() checks the distance from the mean in both directions.
- Values with an absolute Z-score above 3 are identified as potential outliers.

## **5. Percentile Method**



**Understand the Concept**

- The Percentile Method identifies extreme values using selected percentile limits.
- Values below the lower percentile or above the upper percentile are treated as potential outliers.


**Demonstrate the Concept**

**Example:**

- Use the 1st and 99th percentiles to find extreme study hour values.

**AI/ML Usage:**

- Helps identify extreme values before model training.

```python
# Implement the Concept
# percentiles
lower = df["Study_Hours"].quantile(0.01)
upper = df["Study_Hours"].quantile(0.99)

# Find outliers
outliers = df[(df["Study_Hours"] < lower) | (df["Study_Hours"] > upper)]
print("Lower Limit:", lower)
print("Upper Limit:", upper)
print("Outliers:", len(outliers))

```

```python
outliers[["Student_ID", "Study_Hours"]].sample(4)
```

**Explanation**

- quantile() finds the selected percentile values.
- The lower and upper limits define the extreme range.

## **6. Winsorization**



**Understand the Concept**

- Winsorization reduces the effect of extreme values by replacing them with selected percentile limits.
- It keeps the records but changes extreme values to less extreme values.

**Should it be capped?**
- Yes, when valid extreme values are present but have too much influence on the analysis.

**Why?**
- It reduces the impact of extreme values without removing the observations.



**Demonstrate the Concept**

**Example:**

- Replace values below the 1st percentile and above the 99th percentile.

**AI/ML Usage:**

- Helps reduce the influence of extreme values while retaining the data.

```python
# Implement the Concept
# limits
lower = df["Study_Hours"].quantile(0.01)
upper = df["Study_Hours"].quantile(0.99)

# Apply Winsorization
outliers["Study_Hours_Winsorized"] = df["Study_Hours"].clip(lower, upper)

# Compare values
outliers[["Student_ID","Study_Hours","Study_Hours_Winsorized"]].sample(5)
```

**Explanation**

- quantile() finds the lower and upper limits.
- clip() replaces extreme values with those limits

## **7. Capping**



**Understand the Concept**

- Capping limits extreme values by replacing them with a fixed upper or lower boundary.
- It keeps the records while reducing the effect of extreme values.

**Should it be capped?**
- Cap the value when it is valid but has too much influence on the analysis.

**Why?**
- Capping controls extreme values without removing the observations.



**Demonstrate the Concept**

**Example:**

- Limit extreme Study_Hours values to the selected upper and lower limits.

**AI/ML Usage:**

- Helps reduce the effect of extreme values on ML models.

```python
# Implement the Concept
# limits
lower = df["Study_Hours"].quantile(0.01)
upper = df["Study_Hours"].quantile(0.99)

# Apply capping
outliers["Study_Hours_Capped"] = df["Study_Hours"].clip(lower, upper)

# Show results
outliers[["Student_ID","Study_Hours","Study_Hours_Capped"]].sample(5)
```

**Explanation**

- quantile() sets the lower and upper limits.
- clip() replaces values outside the limits.

## **8. Flooring**



**Understand the Concept**

- Flooring replaces values below a selected lower limit with that limit.
- It reduces the effect of unusually low values while keeping the records.


**Demonstrate the Concept**

**Example:**

- Set a minimum limit for Study_Hours and replace lower values.

**AI/ML Usage:**

- Helps reduce the effect of unusually low values on ML models.

```python
# Implement the Concept
# lower limit
lower = df["Study_Hours"].quantile(0.01)

# Apply flooring
outliers["Study_Hours_Floored"] = df["Study_Hours"].clip(lower=lower)

# Show results
outliers[["Student_ID","Study_Hours","Study_Hours_Floored"]].head(10)
```

**Explanation**

- quantile() finds the lower limit.
- clip() replaces values below the limit.

## **10. Removing Outliers**



**Understand the Concept**

- Removing outliers means deleting observations that are confirmed to be incorrect or unsuitable for the analysis.
- Outliers should not be removed automatically because some extreme values may be valid.


**Is it a valid extreme observation?**
- Do not remove it when the extreme value is realistic and meaningful.

**Should it be removed?**
- Remove only when there is a clear reason, such as a data-entry error.

**Why?**
- Removing valid extreme observations can cause information loss and may make the dataset less representative.



**Demonstrate the Concept**

**Example:**

- Remove study hour values that are negative or clearly invalid.

**AI/ML Usage:**

- Helps prevent incorrect extreme values from affecting model training.

```python
# Implement the Concept
# invalid values
invalid = df[df["Study_Hours"] < 0]
print("Invalid Outliers:", len(invalid))

# Remove invalid values
df_clean = df[df["Study_Hours"] >= 0]
print("Rows Before:", len(df))
print("Rows After:", len(df_clean))
```

**Explanation**

- The condition finds negative study hour values.
- These values are treated as errors.
- Only the confirmed invalid records are removed.
