## **1. Why Use Pipelines?**



**Understand the Concept**

- Pipelines combine multiple preprocessing steps into one workflow.
- They make preprocessing easier to manage and reuse.
- They help apply the same steps consistently to training and test data.
- Pipelines also help reduce the risk of data leakage.

**Demonstrate the Concept**

**Example:**

- Combine imputation and scaling for numerical features.

**AI/ML Usage:**

- Helps create a clean and consistent preprocessing workflow before model training.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Numerical preprocessing pipeline
numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),("scaler", StandardScaler())])
print(numeric_pipeline)
```

**Explanation**

- SimpleImputer fills missing numerical values.
- StandardScaler scales the numerical features.
- Pipeline combines both steps into one workflow.

## **2. Pipeline**



**Understand the Concept**

- Pipeline connects multiple preprocessing steps in a fixed order.
- Each step performs a specific task on the data.
- It keeps the preprocessing workflow simple and organized.
- It helps apply the same transformations consistently to new data.



**Demonstrate the Concept**

**Example:**

- Create a pipeline that fills missing values and then scales numerical features.

**AI/ML Usage:**

- Helps automate multiple preprocessing steps before model training.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# numerical pipeline
numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),("scaler", StandardScaler())])
print(numeric_pipeline)
```

**Explanation**

- Pipeline combines the imputation and scaling steps.
- The steps are applied in the given order.
- Missing values are handled first, followed by scaling.

## **3. ColumnTransformer**



**Understand the Concept**

- `ColumnTransformer` applies different preprocessing steps to different columns.
- It is useful when a dataset has both numerical and categorical features.
- It allows multiple preprocessing pipelines to work together.



**Demonstrate the Concept**

**Example:**

- Apply scaling to numerical columns and encoding to categorical columns.

**AI/ML Usage:**

- Helps preprocess different types of features in one workflow.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = ["Age","Monthly_Income","Years_At_Company","Job_Satisfaction"]

categorical_features = ["Overtime","Department","Job_Role","Education_Level"]

preprocessor = ColumnTransformer([("num", StandardScaler(), numeric_features),("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)])
print(preprocessor)
```

**Explanation**

- Numerical columns are processed using `StandardScaler`.
- Categorical columns are processed using `OneHotEncoder`.
- `ColumnTransformer` combines both transformations.

## **4. Numerical Pipeline**



**Understand the Concept**

- A numerical pipeline handles numerical features.
- Missing numerical values can be filled using imputation.
- The values can then be scaled for machine learning.



**Demonstrate the Concept**

**Example:**

- Apply median imputation and standard scaling to numerical features.

**AI/ML Usage:**

- Helps prepare numerical data for machine learning models.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")), ("scaler", StandardScaler())])
print(numeric_pipeline)
```

**Explanation**

- `SimpleImputer` fills missing numerical values using the median.
- `StandardScaler` scales the numerical features.
- Both steps are combined in one pipeline.

## **5. Categorical Pipeline**



**Understand the Concept**

- A categorical pipeline handles categorical features.
- Missing categories can be replaced with the most frequent value.
- Categorical values can then be converted into numerical form.



**Demonstrate the Concept**

**Example:**

- Apply most-frequent imputation and one-hot encoding to categorical features.

**AI/ML Usage:**

- Helps convert categorical data into a format that machine learning models can use.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                                 ("encoder", OneHotEncoder(handle_unknown="ignore"))])
print(categorical_pipeline)
```

**Explanation**

- `SimpleImputer` fills missing categorical values.
- `OneHotEncoder` converts categories into numerical features.
- Both steps are combined into one pipeline.

## **7. Scaling Pipeline**



**Understand the Concept**

- Scaling brings numerical features to a similar range.
- It can help models that are sensitive to feature scale.
- Scaling is usually performed after handling missing values.



**Demonstrate the Concept**

**Example:**

- Apply StandardScaler to numerical employee features.

**AI/ML Usage:**

**Explanation**

## **8. Encoding Pipeline**



**Understand the Concept**

- Encoding converts categorical values into numerical values.
- One-hot encoding creates separate features for each category.
- `handle_unknown="ignore"` allows the pipeline to handle new categories.



**Demonstrate the Concept**

**Example:**

- Apply one-hot encoding to employee categorical features.

**AI/ML Usage:**

- Helps machine learning models work with categorical data.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

encoding_pipeline = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                              ("encoder", OneHotEncoder(handle_unknown="ignore"))])
print(encoding_pipeline)
```

**Explanation**

- Missing categorical values are filled first.
- `OneHotEncoder` converts categories into numerical features.
- `handle_unknown="ignore"` prevents errors from unseen categories.

## **9. Combining Multiple Transformations**



**Understand the Concept**

- Different preprocessing steps can be combined into one workflow.
- Numerical and categorical features can use different transformations.
- `ColumnTransformer` combines these pipelines together.



**Demonstrate the Concept**

**Example:**

- Combine numerical imputation and scaling with categorical imputation and encoding.

**AI/ML Usage:**

- Helps prepare the complete dataset in a single preprocessing workflow.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

numeric_features = ["Age", "Monthly_Income", "Years_At_Company","Job_Level", "Job_Satisfaction", "Work_Hours_Per_Week"]

categorical_features = ["Overtime", "Department", "Job_Role","Education_Level", "Marital_Status"]

numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),
                             ("scaler", StandardScaler())])

categorical_pipeline = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                                 ("encoder", OneHotEncoder(handle_unknown="ignore"))])

preprocessor = ColumnTransformer([("num", numeric_pipeline, numeric_features),
                                  ("cat", categorical_pipeline, categorical_features)])
print(preprocessor)
```

**Explanation**

- Numerical features are imputed and scaled.
- Categorical features are imputed and encoded.

## **10. `fit()`**



**Understand the Concept**

- `fit()` learns the required values from the training data.
- For example, it learns the median for imputation and values needed for scaling.
- It does not return the transformed data.



**Demonstrate the Concept**

**Example:**

- Fit the complete preprocessing pipeline using training data.

**AI/ML Usage:**

- Helps learn preprocessing parameters from training data only.

```python
# features and target
X = df.drop(columns=["Employee_ID", "Attrition"])
y = df["Attrition"]

# Fit the preprocessing pipeline
preprocessor.fit(X)
```

**Explanation**

- `fit()` learns the preprocessing parameters from `X`.
- These learned parameters can be used to transform the data.

## **11. `transform()`**



**Understand the Concept**

- `transform()` applies the parameters already learned by `fit()`.
- It converts the data into the required processed format.
- It is commonly used on validation and test data.



**Demonstrate the Concept**

**Example:**

- Transform the feature data using the fitted preprocessing pipeline.

**AI/ML Usage:**

- Helps apply the same preprocessing rules to new data.

```python
# Transform the data using the fitted pipeline
X_transformed = preprocessor.transform(X)
print("Transformed data shape:", X_transformed.shape)
```

**Explanation**

- `transform()` applies the learned preprocessing steps.
- It returns the processed feature data.

## **12. `fit_transform()`**



**Understand the Concept**

- `fit_transform()` combines `fit()` and `transform()` in one step.
- It first learns the preprocessing parameters.
- It then transforms the same data.



**Demonstrate the Concept**

**Example:**

- Learn and apply all preprocessing steps to the training features.

**AI/ML Usage:**

- Provides a simple way to fit and transform training data.

```python
# Fit and transform the feature data
X_train_transformed = preprocessor.fit_transform(X)

print("Transformed training data shape:", X_train_transformed.shape)
```

**Explanation**

- `fit_transform()` learns the required parameters and transforms the data.
- It is commonly used for training data.
