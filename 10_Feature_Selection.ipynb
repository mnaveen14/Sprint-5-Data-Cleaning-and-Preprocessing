```python
import pandas as pd 
df = pd.read_csv("used_car_dataset.csv")
df.info()
```

## **1. What is Feature Selection?**



**Understand the Concept**

- Feature selection means choosing the most useful features from a dataset.
- It removes unnecessary, irrelevant, or repeated features.

**Demonstrate the Concept**

**Example:**
- Select useful features for predicting Selling_Price_Lakhs.

**AI/ML Usage:**
- Helps models focus on important information and reduce unnecessary data.

## **2. Why Feature Selection?**



**Understand the Concept**

- Feature selection removes features that do not provide useful information.
- Unnecessary features can add noise and increase model complexity.
- Too many unnecessary features can increase training time and cause overfitting.

**Demonstrate the Concept**

**Example:**
- Remove unwanted col because it has the same value for every record.



## **3. Relevant Features**



**Understand the Concept**

- Relevant features have useful information related to the target variable.
- They can help the machine learning model make better predictions.
- Selecting relevant features can improve model performance.

**Demonstrate the Concept**

**Example:**
- Use Mileage_Km as a relevant feature for predicting Selling_Price_Lakhs.

**AI/ML Usage:**
- Relevant features help models learn meaningful patterns from the data.

```python
# Implement the Concept
# relevant feature
relevant_feature = df[["Mileage_Km", "Selling_Price_Lakhs"]]
print(relevant_feature.head())
```

**Explanation**

- Mileage_Km contains useful information for predicting Selling_Price_Lakhs.
- The feature and target are selected together for analysis.

## **4. Irrelevant Features**



**Understand the Concept**

- Irrelevant features do not provide useful information for predicting the target.
- They can add noise and make the model more complex.
- Removing them can help the model focus on useful information.

**Demonstrate the Concept**

**Example:**
- Remove Random_Feature because it does not provide useful information for predicting Selling_Price_Lakhs.

**AI/ML Usage:**
- Removing irrelevant features can improve model efficiency and generalization.

```python
# Implement the Concept
# Remove the irrelevant feature
df = df.drop(columns=["Random_Feature"])
print(df.columns)
```

**Explanation**

- Random_Feature is not useful for predicting the target.
- Removing it reduces unnecessary information.

## **5. Redundant Features**



**Understand the Concept**

- Redundant features contain information that is already provided by another feature.
- They may provide duplicate or highly similar information.
- Removing redundant features can reduce unnecessary complexity.

**Demonstrate the Concept**

**Example:**
- Remove Engine_Power_Copy because it contains the same information as Horsepower.

**AI/ML Usage:**
- Helps reduce duplicate information and improve model efficiency.

```python
# Implement the Concept
# Check whether the two features are identical
print((df["Horsepower"] == df["Engine_Power_Copy"]).all())
```

```python
# Remove the redundant feature
df = df.drop(columns=["Engine_Power_Copy"])
print(df.columns)
```

**Explanation**

- The comparison checks whether Horsepower and Engine_Power_Copy contain the same values.
- Engine_Power_Copy is redundant because it duplicates Horsepower.
- Removing it avoids duplicate information.

## **6. Correlation-Based Selection**



**Understand the Concept**

- Correlation measures the relationship between numerical features and the target.
- A higher absolute correlation can indicate a stronger relationship with the target.
- Correlation can help identify useful numerical features.

**Demonstrate the Concept**

**Example:**
- Find the correlation between numerical features and Selling_Price_Lakhs.

**AI/ML Usage:**
- Helps identify numerical features that may be useful for prediction.

```python
# Implement the Concept
# correlations 
correlation = df.select_dtypes(include="number").corr()["Selling_Price_Lakhs"]
print(correlation.sort_values(ascending=False))
```

**Explanation**

- corr() calculates the correlation between numerical features.
- The result shows how strongly each feature is related to Selling_Price_Lakhs.

## **7. Variance Threshold**



**Understand the Concept**

- Variance measures how much a feature changes across records.
- A feature with very low variance provides little information.
- Variance Threshold can remove features with low variance automatically.

**Demonstrate the Concept**

**Example:**
- Identify features with very low variance.

**AI/ML Usage:**
- Helps remove features that provide little variation to the model.

```python
# Implement the Concept
from sklearn.feature_selection import VarianceThreshold

# numerical features
X = df.drop(columns=["Selling_Price_Lakhs"]).select_dtypes(include="number")

# Variance Threshold
selector = VarianceThreshold(threshold=0.01)
selector.fit(X)
selected_features = X.columns[selector.get_support()]
print(selected_features)
```

**Explanation**

- VarianceThreshold checks the variation in each feature.
- Features below the selected threshold are removed.
- The remaining features have enough variation for analysis.

## **8. Univariate Feature Selection**



**Understand the Concept**

- Univariate feature selection evaluates each feature separately against the target.
- Statistical tests are used to measure the usefulness of features.
- The best-scoring features can be selected for the model.

**Demonstrate the Concept**

**Example:**
- Select the top numerical features using SelectKBest.

**AI/ML Usage:**
- Helps reduce the number of features before model training.

```python
# Implement the Concept
from sklearn.feature_selection import SelectKBest, f_regression

X = df.select_dtypes(include="number").drop(columns=["Selling_Price_Lakhs"])
y = df["Selling_Price_Lakhs"]

# Select 5 features
selector = SelectKBest(score_func=f_regression, k=5)
X_selected = selector.fit_transform(X, y)
print(X.columns[selector.get_support()])
```

**Explanation**

- SelectKBest evaluates each feature independently.
- f_regression measures the relationship between features and the numerical target.

## **9. Mutual Information**



**Understand the Concept**

- Mutual Information measures how much information a feature provides about the target.
- It can detect both linear and non-linear relationships.
- Higher mutual information indicates a stronger dependency between a feature and the target.

**Demonstrate the Concept**

**Example:**
- Calculate Mutual Information scores for numerical features and Selling_Price_Lakhs.

**AI/ML Usage:**
- Helps identify features with useful linear or non-linear relationships.

```python
# Implement the Concept
from sklearn.feature_selection import mutual_info_regression

X = df.select_dtypes(include="number").drop(columns=["Selling_Price_Lakhs"])
y = df["Selling_Price_Lakhs"]

# Mutual Information scores
scores = mutual_info_regression(X, y, random_state=42)
mi_scores = pd.Series(scores, index=X.columns)
print(mi_scores.sort_values(ascending=False))
```

**Explanation**

- Mutual Information measures the dependency between each feature and the target.
- Higher scores indicate more useful relationships.
- It can identify non-linear relationships that correlation may miss.

## **10. Recursive Feature Elimination**



**Understand the Concept**

- Recursive Feature Elimination selects features by repeatedly removing less important features.
- It uses a machine learning model to rank feature importance.
- The process continues until the required number of features remains.

**Demonstrate the Concept**

**Example:**
- Select the top 5 features using Recursive Feature Elimination.

**AI/ML Usage:**
- Helps create a smaller feature set based on model performance.

```python
# Implement the Concept
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestRegressor

# features and target
X = df.select_dtypes(include="number").drop(columns=["Selling_Price_Lakhs"])
y = df["Selling_Price_Lakhs"]

# Apply RFE
model = RandomForestRegressor(n_estimators=50, random_state=42)
selector = RFE(model, n_features_to_select=5)
selector.fit(X, y)
print(X.columns[selector.support_])
```

**Explanation**

- RFE starts with all selected features.
- Less important features are removed step by step.

## **11. Feature Importance**



**Understand the Concept**

- Feature importance shows how much each feature contributes to a model's predictions.
- Tree-based models can calculate feature importance.
- Important features can be prioritized during feature selection.

**Demonstrate the Concept**

**Example:**
- Calculate feature importance using a Random Forest model.

**AI/ML Usage:**
- Helps identify which features contribute most to model predictions.

```python
# Implement the Concept
from sklearn.ensemble import RandomForestRegressor

X = df.select_dtypes(include="number").drop(columns=["Selling_Price_Lakhs"])
y = df["Selling_Price_Lakhs"]

# Train Random Forest
model = RandomForestRegressor(n_estimators=50, random_state=42)
model.fit(X, y)

# feature importance
importance = pd.Series(model.feature_importances_,index=X.columns).sort_values(ascending=False)
print(importance)
```

**Explanation**

- Random Forest calculates an importance score for each feature.
- Higher scores indicate greater contribution to the model.
- These scores can help decide which features are useful.
