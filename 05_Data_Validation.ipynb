## **2. Range Validation**



**Understand the Concept**

- Range validation checks whether a numerical value falls within an allowed range.
- It helps identify values that are too high, too low, or outside the expected limits.



**Demonstrate the Concept**

**Example:**

- Age should be between 18 and 100.
- Discount Percentage should be between 0 and 100.

**AI/ML Usage:**

- Helps prevent extreme or invalid values from affecting analysis.

**Why is it required?**

- To detect invalid numerical values.
- To ensure values follow expected business ranges.

```python
# Convert Age to numeric
age = pd.to_numeric(df["Age"], errors="coerce")

# Convert discount to numeric
discount = pd.to_numeric(df["Discount_Percentage"], errors="coerce")
```

```python
# Implement the Concept
# Check age range
invalid_age = df[(age < 18) | (age > 100)]

# Check discount range
invalid_discount = df[(discount < 0) | (discount > 100)]
print("Invalid Age:", len(invalid_age))
print("Invalid Discount:", len(invalid_discount))

# Invalid age
invalid_age[["Customer_ID", "Age"]].head()
```

**Explanation**

- to_numeric() converts valid values into numbers.
- errors="coerce" converts values like Unknown into missing values.
- This allows range conditions to work without a TypeError.

## **3. Type Validation**



**Understand the Concept**

- Type validation checks whether each column contains the expected data type.
- It helps find values stored in the wrong format, such as text in a numerical column.



**Demonstrate the Concept**

**Example:**

- Age should be numeric.
- Quantity should be numeric.
- Order_Date should be a date.

**AI/ML Usage:**

- Ensures features have the correct data types.
- Prevents errors during data processing and model training.

**Why is it required?**

- To identify incorrect data types.
- To make sure columns are ready for analysis and calculations.

```python
# Implement the Concept
# data types
print(df[["Age", "Quantity", "Unit_Price", "Order_Date"]].dtypes)

```

```python
# invalid numeric values
age_check = pd.to_numeric(df["Age"], errors="coerce")
invalid_age = df[age_check.isna()]
print("Invalid Age:", len(invalid_age))

```

```python
# invalid values
invalid_age[["Customer_ID", "Age"]].head()
```

**Explanation**

- dtypes shows the current data types.
- to_numeric() converts valid values to numbers.
- errors="coerce" changes invalid values to missing values.

## **4. Format Validation**



**Understand the Concept**

- Format validation checks whether values follow the expected pattern.
- It is commonly used for emails, phone numbers, IDs, and dates.



**Demonstrate the Concept**

**Example:**

- Email should contain a valid email format.
- Phone should contain a valid 10-digit number.
- Customer_ID should follow the expected ID format.

**AI/ML Usage:**

- Helps prevent incorrectly formatted data from entering the analysis.
- Improves the quality of customer information used in ML projects.

**Why is it required?**

- To detect incorrectly formatted values.
- To keep data consistent and reliable.

```python
# Implement the Concept
# Check email format
email_check = df["Email"].str.match( r"^[\w\.-]+@[\w\.-]+\.\w+$", na=False)

# Check phone format
phone_check = df["Phone"].str.match(r"^\d{10}$",na=False)
print("Invalid Emails:", (~email_check).sum())
print("Invalid Phones:", (~phone_check).sum())

```

```python
# invalid emails
df.loc[~email_check, ["Customer_ID", "Email"]].head()
```

**Explanation**

- str.match() checks whether values follow a pattern.
- The email and phone patterns identify invalid formats.

## **5. Category Validation**



**Understand the Concept**

- Category validation checks whether values belong to an accepted set of categories.
- It helps identify unexpected or invalid category values.



**Demonstrate the Concept**

**Example:**

- Gender can be `Male`, `Female`, or `Other`.
- Payment_Method can be `UPI`, `Credit Card`, `Debit Card`, `Net Banking`, or `Cash`.

**Real-World Example:**

- A Gender value such as "Unknown" is not an accepted category.
- A Payment_Method such as "Crypto" is invalid if it is not supported.

**AI/ML Usage:**

- Keeps categorical features consistent.
- Prevents unexpected categories from causing problems during preprocessing and model training.


```python
# Implement the Concept
# Set valid categories
valid_gender = ["Male", "Female", "Other"]
# invalid categories
invalid_gender = df[~df["Gender"].isin(valid_gender)]
print("Invalid Gender:", len(invalid_gender))
```

```python
# invalid values
invalid_gender[["Customer_ID", "Gender"]].head()
```

**Explanation**

- A list contains the accepted categories.
- isin() checks whether values belong to that list.
- Values outside the list are identified as invalid.

## **7. Business Rule Validation**



**Understand the Concept**

- Business rule validation checks whether data follows rules defined by the business.
- These rules are based on how the data should work in a real-world situation.



**Demonstrate the Concept**

**Example:**

- Order_Amount should match Quantity, Unit_Price, and Discount_Percentage.
- Delivery_Date should not be before Order_Date.
- Quantity should be greater than 0.

**AI/ML Usage:**

- Helps ensure business data is logically correct.
- Prevents incorrect records from affecting analysis and model training.

**Why is it required?**

- To detect logically incorrect records.
- To ensure data follows real-world business rules.

```python
# Convert columns
quantity = pd.to_numeric(df["Quantity"], errors="coerce")
price = pd.to_numeric(df["Unit_Price"], errors="coerce")
discount = pd.to_numeric(df["Discount_Percentage"], errors="coerce")
amount = pd.to_numeric(df["Order_Amount"], errors="coerce")
```

```python
# Implement the Concept
# expected amount
expected = quantity * price * (1 - discount / 100)

# rule violations
invalid = df[(amount - expected).abs() > 1]
print("Invalid Order Amount:", len(invalid))
```

```python
# invalid records
invalid[["Customer_ID", "Quantity", "Unit_Price","Discount_Percentage", "Order_Amount"]].head()
```

**Explanation**

- The expected amount is calculated from the order details.
- It is compared with the actual Order_Amount.

## **8. Referential Validation**



**Understand the Concept**

- Referential validation checks whether values correctly match a valid reference list or dataset.
- It helps identify values that do not exist in the expected reference data.



**Demonstrate the Concept**

**Real-World Example:**

- A customer record contains Country_Code `XX`, but `XX` is not a valid country code.
- The invalid code can cause problems when grouping customers by country.

**AI/ML Usage:**

- Keeps reference-based features accurate and consistent.
- Prevents invalid categories from entering ML models.

**Why is it required?**

- To check values against valid reference data.
- To identify records with invalid references.

```python
# Implement the Concept
# Set valid codes
valid_codes = ["IN", "US", "UK", "CA", "AU", "SG", "AE"]

# invalid codes
invalid = df[~df["Country_Code"].isin(valid_codes)]
print("Invalid Country Codes:", len(invalid))
```

```python
# invalid records
invalid[["Customer_ID", "Country_Code"]].head()
```

**Explanation**

- valid_codes contains the accepted reference values.
- isin() checks each Country_Code against the list.
- Values not found in the list are identified as invalid.

## **9. Date Validation**


**Understand the Concept**

- Date validation checks whether date values are valid and follow the expected format.
- It also checks whether dates follow logical rules.



**Demonstrate the Concept**

**Example:**

- Order_Date should be a valid date.
- Delivery_Date should be a valid date.
- Delivery_Date should not be before Order_Date.

**AI/ML Usage:**

- Prevents incorrect dates from affecting time-based analysis.

**Why is it required?**

- To identify invalid date values.
- To check the logical order of dates.

```python
# Implement the Concept
# Convert dates
order_date = pd.to_datetime(df["Order_Date"], errors="coerce")
delivery_date = pd.to_datetime(df["Delivery_Date"], errors="coerce")

# invalid dates
invalid_date = df[order_date.isna() | delivery_date.isna()]
print("Invalid Dates:", len(invalid_date))
```

```python
# date order
invalid_order = df[delivery_date.notna() & order_date.notna() & (delivery_date < order_date)]
print("Invalid Date Order:", len(invalid_order))
```

```python
# invalid records
invalid_order[["Customer_ID", "Order_Date", "Delivery_Date"]].head()
```

**Explanation**

- to_datetime() checks whether values are valid dates.
- errors="coerce" converts invalid dates to missing values.
- The date comparison finds deliveries before the order date.

## **10. Constraint Validation**



**Understand the Concept**

- Constraint validation checks whether data follows specific conditions or restrictions.
- It helps ensure that values and relationships between columns remain valid.



**Demonstrate the Concept**

**Example:**

- Customer_ID should be unique.
- Quantity should be greater than 0.
- Discount_Percentage should be between 0 and 100.
- Customer_Status should contain only accepted values.

**AI/ML Usage:**

- Helps maintain consistent and reliable data.
- Prevents invalid records from affecting analysis and model training.



```python
# Implement the Concept
# duplicate IDs
duplicate_id = df[df["Customer_ID"].duplicated(keep=False)]

# status
valid_status = ["Active", "Inactive"]
invalid_status = df[~df["Customer_Status"].isin(valid_status)]
print("Duplicate IDs:", len(duplicate_id))
print("Invalid Status:", len(invalid_status))
```

```python
# invalid status 
invalid_status[["Customer_ID", "Customer_Name","Customer_Status"]].head()
```

**Explanation**

- isin() checks whether Customer_Status follows the allowed values.
