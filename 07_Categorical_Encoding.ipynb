# **Notebook 7 – Categorical Encoding**

```python
import pandas as pd
df = pd.read_csv("customer_churn_dataset.csv")
df.info()
```

## **1. What is Categorical Encoding?**

**Understand the Concept**

- Categorical encoding converts categories into numerical values.
- The encoding method should match the type of categorical variable.

**When should each encoding technique be used?**

- Label Encoding: Simple or binary categories.
- Ordinal Encoding: Categories with a clear order.
- One-Hot Encoding: Nominal categories with few values.
- Dummy Variables: When one category should be removed.
- Frequency Encoding: Categories with many values.
- Target Encoding: High-cardinality categories with a target.

**Risks of inappropriate encoding**

- Wrong encoding can create false relationships between categories.
- It can reduce model performance or cause data leakage.

## **2. Nominal Variables**



**Understand the Concept**

- Nominal variables are categories with no natural order.
- Examples include Gender, City, and Payment_Method.
- The categories are different, but one category is not higher or lower than another.

**Demonstrate the Concept**

- Gender has categories such as Male, Female, and Other.
- These categories do not have any ranking.

```python
# Implement the Concept
# nominal variable
print(df["Gender"].unique())
```

```python
# gender counts
print(df["Gender"].value_counts())
```

**Explanation**

- unique() shows the available categories.
- value_counts() shows how often each category appears.
- Nominal categories should not be given an artificial order.

## **3. Ordinal Variables**



**Understand the Concept**

- Ordinal variables are categories that have a meaningful order.
- The difference between categories is not necessarily equal.
- Examples include Education_Level and Satisfaction_Level.

**Demonstrate the Concept**

**Example:**
- Use Satisfaction_Level by preserving the order of its categories.

**AI/ML Usage:**
- Helps models understand the ranking between categories.

```python
# Implement the Concept
# ordinal variable
print(df["Satisfaction_Level"].unique())

# category counts
print(df["Satisfaction_Level"].value_counts())
```

**Explanation**

- Satisfaction_Level has a natural order.
- value_counts() shows the number of records in each category.

## **4. Binary Variables**



**Understand the Concept**

- Binary variables have only two possible categories.
- They are commonly used for Yes/No or True/False information.

**Demonstrate the Concept**

**Example:**
- Identify Is_Member as a binary variable.

**AI/ML Usage:**
- Binary variables can be converted into numerical values for machine learning models.

```python
# Implement the Concept
# Is_Member categories
print(df["Is_Member"].unique())
```

**Explanation**

- unique() shows the categories in Is_Member.
- This confirms that the variable has two categories.

## **5. Label Encoding**



**Understand the Concept**

- Label Encoding converts each category into a numerical label.
- It is mainly suitable for binary variables or categories where an order is not important.
- For nominal variables with many categories, other encoding methods may be more suitable.

**Demonstrate the Concept**

**Example:**
- Apply Label Encoding to Is_Member.

**AI/ML Usage:**
- Converts categorical values into numerical values that models can process.

```python
# Implement the Concept
from sklearn.preprocessing import LabelEncoder
# label encoder
encoder = LabelEncoder()
df["Is_Member_Encoded"] = encoder.fit_transform(df["Is_Member"])
df[["Is_Member", "Is_Member_Encoded"]].head()
```

**Explanation**

- LabelEncoder assigns a numerical label to each category.
- fit_transform() learns the categories and converts them into numbers.

## **6. Ordinal Encoding**



**Understand the Concept**

- Ordinal Encoding converts ordered categories into numerical values.
- It keeps the natural order between categories.
- It is useful when categories have a meaningful ranking.

**Demonstrate the Concept**

**Example:**
- Apply Ordinal Encoding to Satisfaction_Level.

**AI/ML Usage:**
- Helps models understand the order between categories.

```python
# Implement the Concept
from sklearn.preprocessing import OrdinalEncoder
# Ordinal encoder
encoder = OrdinalEncoder(categories=[["Low", "Medium", "High"]])
df["Satisfaction_Level_Encoded"] = encoder.fit_transform(df[["Satisfaction_Level"]])
df[["Satisfaction_Level", "Satisfaction_Level_Encoded"]].head()
```

**Explanation**

- OrdinalEncoder converts ordered categories into numbers.
- categories defines the correct order.

## **7. One-Hot Encoding**



**Understand the Concept**

- One-Hot Encoding creates a separate column for each category.
- Each category is represented using 0 or 1.
- It is mainly used for nominal variables with a small number of categories.

**Demonstrate the Concept**

**Example:**
- Apply One-Hot Encoding to Payment_Method.

**AI/ML Usage:**
- Helps models use nominal categories without creating an artificial order.

```python
# Implement the Concept
from sklearn.preprocessing import OneHotEncoder
# One-hot encoder
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(df[["Payment_Method"]])
encoded_df = pd.DataFrame(encoded,columns=encoder.get_feature_names_out(["Payment_Method"]))
encoded_df.head()
```

**Explanation**

- OneHotEncoder creates a separate column for each category.
- Each category is represented using 0 or 1.

## **8. Dummy Variables**



**Understand the Concept**

- Dummy Variables represent categories using 0 and 1.
- One category is usually removed to avoid redundant information.
- They are commonly used for nominal variables.

**Demonstrate the Concept**

**Example:**
- Create Dummy Variables for Product_Category.

**AI/ML Usage:**
- Helps models use categorical data without unnecessary redundant columns.

```python
# Implement the Concept
# dummy variables
dummy_df = pd.get_dummies(df["Product_Category"], prefix="Product", drop_first=True)
dummy_df.head()
```

**Explanation**

- get_dummies() creates 0 and 1 columns for categories.
- drop_first=True removes one category.
- Removing one category helps avoid redundant information.

## **9. Frequency Encoding**



**Understand the Concept**

- Frequency Encoding replaces each category with its frequency in the dataset.
- It is useful for categorical variables with many unique categories.
- It reduces the number of new columns compared with One-Hot Encoding.

**Demonstrate the Concept**

**Example:**
- Apply Frequency Encoding to City.

**AI/ML Usage:**
- Helps models use high-cardinality categories with fewer features.

```python
# Implement the Concept
# category frequency
frequency = df["City"].value_counts()

# frequency encoding
df["City_Frequency"] = df["City"].map(frequency)
df[["City", "City_Frequency"]].head()
```

**Explanation**

- value_counts() calculates how often each city appears.
- map() replaces each city with its frequency.
- The new column contains numerical frequency values.

## **10. Target Encoding**



**Understand the Concept**

- Target Encoding replaces each category with the average value of the target variable.
- It is useful for categorical variables with many unique categories.
- It should be applied carefully to avoid data leakage.

**Demonstrate the Concept**

**Example:**
- Apply Target Encoding to City using Churn as the target.

**AI/ML Usage:**
- Helps models capture the relationship between categories and the target.

```python
# Implement the Concept
# target mean
target_mean = df.groupby("City")["Churn"].apply(lambda x: (x == "Yes").mean())

# target encoding
df["City_Target_Encoded"] = df["City"].map(target_mean)
df[["City", "City_Target_Encoded"]].head()
```

**Explanation**

- groupby() calculates the target rate for each city.
- map() assigns the target rate to each city.
- Target Encoding should be calculated carefully to prevent data leakage.

## **11. Handling Unknown Categories**



**Understand the Concept**

- Unknown categories are values that were not present during training.
- The encoding method should handle new categories without causing errors.
- This is important when new data contains categories not seen before.

**Demonstrate the Concept**

**Example:**
- Handle a new Contract_Type category that was not present during training.

**AI/ML Usage:**
- Helps models process new categories safely.

```python
# Implement the Concept
from sklearn.preprocessing import OneHotEncoder

# One-hot encoder
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

# Fit encoder
encoder.fit(df[["Contract_Type"]])
new_data = pd.DataFrame({"Contract_Type": ["Biannual"]})
encoded = encoder.transform(new_data)
print(encoded)
```

**Explanation**

- handle_unknown="ignore" prevents errors for new categories.
- The unseen category is encoded without creating a new column.

## **12. High-Cardinality Categories**

**Understand the Concept**

- High-cardinality categories have a large number of unique values.
- One-Hot Encoding can create too many columns for these categories.
- Frequency Encoding or Target Encoding can be more suitable.

**Demonstrate the Concept**

**Example:**
- Use City as a high-cardinality categorical variable.

**AI/ML Usage:**
- Helps reduce the number of features while keeping useful category information.

```python
# Implement the Concept
# Count unique categories
unique_cities = df["City"].nunique()
print("Unique Cities:", unique_cities)
```

```python
# Frequency Encoding
frequency = df["City"].value_counts()
df["City_Frequency"] = df["City"].map(frequency)

# Target Encoding
target_mean = df.groupby("City")["Churn"].apply(lambda x: (x == "Yes").mean())
df["City_Target_Encoded"] = df["City"].map(target_mean)
df[["City","City_Frequency","City_Target_Encoded"]].head()
```

**Explanation**

- nunique() counts the number of unique cities.
- A large number of categories can create too many features.
- Frequency Encoding replaces each city with its frequency.
- Target Encoding replaces each city with its Churn rate.
- Both methods handle high-cardinality categories with fewer columns.
