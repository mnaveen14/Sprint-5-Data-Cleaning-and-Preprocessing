# **Notebook 3 – Missing Value Handling**

```python
import pandas as pd
df = pd.read_csv("house_prices_dataset.csv")
df.info()
```

## **1. What are Missing Values?**



**Understand the Concept**

- Missing values are data points that are not available in a dataset.
- They can appear as `NaN`, `None`, or empty values.

**When to use it**
- Check for missing values before data analysis or ML.

**When not to use it**
- Do not handle all missing values using the same method.

**Advantages**
- Helps identify incomplete data.
- Makes data preparation easier.

**Limitations**
- Missing data can affect analysis.
- Different missing patterns may need different methods.



**Demonstrate the Concept**

**Real-World Example:**
- A house dataset may have missing area, price, or location.

**AI/ML Usage:**
- Missing values should be handled before training a house price prediction model.

**Why is it required?**
- It helps keep the dataset complete and useful.

```python
# Implement the Concept
print("Missing Values:")
print(df.isnull().sum())
```

**Explanation**

- `isnull()` checks for missing values.
- `sum()` counts the missing values.

## **2. Why Missing Values Occur?**



**Understand the Concept**

- Missing values can happen when data is not collected or entered properly.
- They can also occur because of system errors, unavailable information, or skipped fields.

**When to use it**
- Check why the data is missing before choosing a handling method.

**When not to use it**
- Do not assume all missing values have the same reason.

**Advantages**
- Helps choose a suitable missing-value treatment.
- Gives a better understanding of the dataset.

**Limitations**
- The exact reason for missing data may not always be known.
- Finding the reason can take extra time.



**Demonstrate the Concept**

**Real-World Example:**
- Some property records may not have parking or year-built information.

**AI/ML Usage:**
- Understanding why values are missing helps select a suitable preprocessing method.

## **3. MCAR – Missing Completely At Random**



**Understand the Concept**

- MCAR means the missing value has no connection with any data in the dataset.
- The missing values happen completely randomly.
- Any record has an equal chance of having a missing value.

**When to use it**
- When missing values are truly random.

**When not to use it**
- When the missing value is related to another column or the missing information itself.

**Advantages**
- Missing values are usually easier to handle.
- Dropping or simple imputation may work well.

**Limitations**
- True MCAR is uncommon in real-world datasets.
- Removing many rows can reduce the dataset size.



**Demonstrate the Concept**

**Real-World Example:**
- A few house records randomly miss information without any specific pattern.

**AI/ML Usage:**
- MCAR values can often be handled using suitable imputation or row removal methods.

```python
# Implement the Concept
missing_parking = df[df["Parking_Spaces"].isnull()][["House_ID", "Parking_Spaces"]]

print("Missing Parking Records:")
print(missing_parking.sample(3))

print("\nTotal Missing Values:")
print(df["Parking_Spaces"].isnull().sum())
```

**Explanation**

- `isnull()` finds missing values in `Parking_Spaces`.
- Filtering shows the records where values are missing.

## **4. MAR – Missing At Random**



**Understand the Concept**

- MAR means the missing value is related to other information in the dataset.
- The missing value is not directly related to the missing information itself.
- We can often use other columns to understand or estimate the missing value.

**When to use it**
- When missing values depend on other available columns.

**When not to use it**
- When the missing value depends on the missing information itself.

**Advantages**
- Other columns can help estimate the missing values.
- More accurate than simple methods in some cases.

**Limitations**
- It needs useful related columns.
- The relationship between variables may not always be clear.



**Demonstrate the Concept**

**Example:**
- `Price_Lakhs` is missing, but other columns like `Area_sqft` and `Bedrooms` are available.

**AI/ML Usage:**
- Other features can be used to estimate missing values.

```python
# Implement the Concept
missing_price = df[df["Price_Lakhs"].isnull()][["House_ID", "Area_sqft", "Bedrooms", "Price_Lakhs"]]
print(missing_price.head())

print("\nTotal Missing Values:")
print(len(missing_price))
```

**Explanation**

- isnull() finds missing values.
- The selected columns show the required records.
- head() displays the first 5 records.
- len() shows the total number of missing records.

## **5. MNAR – Missing Not At Random**



**Understand the Concept**

- MNAR means the missing value is related to the value that is missing itself.
- People may leave a value blank because of the value they would have entered.
- It is usually harder to identify and handle than MCAR and MAR.

**When to use it**
- When the missing value is likely related to the value itself.

**When not to use it**
- When the missing value is clearly random or related to other known variables.

**Advantages**
- Helps us understand why some important values are missing.
- Can lead to better handling when the reason for missingness is known.

**Limitations**
- The actual reason for missing data may be difficult to know.
- Simple imputation may give misleading results.



**Demonstrate the Concept**

**Real-World Example:**
- People may avoid entering information they consider sensitive or important.

**AI/ML Usage:**
- The missing pattern should be studied carefully before choosing an imputation method.

## **7. Missing Value Percentage**



**Understand the Concept**

- Missing value percentage shows how much data is missing in each column.
- It helps us understand whether the missing values are small or large.

**When to use it**
- Use it when you want to measure the amount of missing data.

**When not to use it**
- Do not use it alone to decide how to handle missing values.

**Advantages**
- Easy to compare missing data across columns.
- Helps identify columns with a high amount of missing data.

**Limitations**
- It does not explain why the values are missing.
- A high percentage does not always mean the column should be removed.



**Demonstrate the Concept**

**Real-World Example:**
- A house dataset may have different amounts of missing information for each column.

**AI/ML Usage:**
- Missing value percentage can help decide whether to impute or remove a column.

```python
# Implement the Concept
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Value Percentage:")
print(missing_percentage.round(2))
```

**Explanation**

- The result shows the missing percentage for each column.

## **8. Dropping Rows**



**Understand the Concept**

- Dropping rows means removing records that contain missing values.
- It is useful when only a small number of rows have missing data.
- We should check the amount of missing data before removing rows.

**When to use it**
- When only a few rows have missing values.
- When the missing records are not important.

**When not to use it**
- When many rows have missing values.
- When removing rows may remove useful information.

**Advantages**
- Simple and easy to apply.
- Does not add estimated values to the dataset.

**Limitations**
- Reduces the number of records.
- Important information may be lost.



**Demonstrate the Concept**

**Example:**
- If only a few house records have missing values, those rows can be removed.

**AI/ML Usage:**
- Removing a small number of incomplete records can help prepare cleaner training data.

```python
# Implement the Concept
print("Rows Before:", len(df))

df_clean = df["Property_Status"].dropna()
print("Rows After:", len(df_clean))
```

**Explanation**

- dropna() removes rows containing missing values.
- len() shows the number of rows before and after removal.

## **9. Dropping Columns**



**Understand the Concept**

- Dropping columns means removing columns that contain too many missing values.
- It can be useful when a column has very little useful information.
- We should check the column before deciding to remove it.

**When to use it**
- When a column has a very high percentage of missing values.
- When the column is not important for analysis or ML.

**When not to use it**
- When the column contains important information.
- When the missing values can be handled using a suitable method.

**Advantages**
- Makes the dataset simpler.
- Removes columns that may not be useful.

**Limitations**
- Important information may be lost.
- Removing a useful feature can affect ML performance.



**Demonstrate the Concept**


**Real-World Example:**
- A property dataset may contain a feature where most values were not collected.

**AI/ML Usage:**
- A column with very high missing values may be removed before model training.

```python

missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Value Percentage:")
print(missing_percentage.round(2))
```

```python
# Implement the Concept
threshold = 50
columns_to_drop = missing_percentage[missing_percentage > threshold].index
df_clean = df.drop(columns=columns_to_drop)

print("\nColumns Removed:")
print(list(columns_to_drop))

print("\nShape Before:", df.shape)
print("Shape After:", df_clean.shape)
```

**Explanation**

- Columns with more than 50% missing values are selected.
- drop() removes the selected columns.

## **10. Mean Imputation**



**Understand the Concept**

- Mean imputation means filling missing numerical values with the average value of that column.
- It is simple and works well when the data is fairly balanced without extreme values.

**When to use it**
- When the column is numerical.
- When the data does not have many outliers.

**When not to use it**
- When the data has many outliers.
- When the data is highly skewed.

**Advantages**
- Simple and easy to apply.
- Keeps the same no of rows.

**Limitations**
- Can reduce the variation in the data.
- Outliers can affect the mean.



**Demonstrate the Concept**


**Real-World Example:**
- Missing house area can be filled with the average area when the values are reasonably distributed.

**AI/ML Usage:**
- Mean imputation can be used to fill missing numerical features before model training.

```python
print("Missing Values Before:")
print(df["Area_sqft"].isnull().sum())
```

```python
# Implement the Concept
df["Area_sqft"] = df["Area_sqft"].fillna(df["Area_sqft"].mean())

print("\nMissing Values After:")
print(df["Area_sqft"].isnull().sum())
```

**Explanation**

- mean() calculates the average value.
- fillna() replaces missing values with the mean.

## **11. Median Imputation**



**Understand the Concept**

- Median imputation means filling missing numerical values with the middle value of the column.
- It works well when the data has outliers or is not evenly distributed.


**When to use it**
- When the column is numerical.
- When the data has outliers or is skewed.

**When not to use it**
- For categorical data.
- When the median does not represent the data well.

**Advantages**
- Less affected by outliers than the mean.
- Simple and easy to apply.

**Limitations**
- Can reduce variation in the data.
- Does not consider relationships between other columns.



**Demonstrate the Concept**


**Real-World Example:**
- Missing house prices can be filled with the median when property prices contain extreme values.

**AI/ML Usage:**
- Median imputation can be used for numerical features before ML model training.

```python
print("Missing Values Before:")
print(df["Price_Lakhs"].isnull().sum())
```

```python
# Implement the Concept
df["Price_Lakhs"] = df["Price_Lakhs"].fillna(df["Price_Lakhs"].median())

print("\nMissing Values After:")
print(df["Price_Lakhs"].isnull().sum())
```

**Explanation**

- median() calculates the middle value.
- fillna() replaces missing values with the median.

## **12. Mode Imputation**



**Understand the Concept**

- Mode imputation means filling missing values with the most frequently occurring value.
- It is mainly used for categorical data.


**When to use it**
- When the column is categorical.
- When only a few values are missing.

**When not to use it**
- When there are too many missing values.
- When the categories have similar frequencies.

**Advantages**
- Simple and easy to use.
- Keeps all the rows in the dataset.

**Limitations**
- Can make one category more common than it should be.
- Does not consider other columns.



**Demonstrate the Concept**


**Real-World Example:**
- Some houses may have a missing Furnishing value.

**AI/ML Usage:**
- Mode imputation can be used to fill missing categorical features before ML training.

```python
print("Missing Values Before:")
print(df["Furnishing"].isnull().sum())
```

```python
# Implement the Concept
df["Furnishing"] = df["Furnishing"].fillna(df["Furnishing"].mode()[0])

print("\nMissing Values After:")
print(df["Furnishing"].isnull().sum())
```

**Explanation**

- mode() finds the most frequent value.
- fillna() replaces missing values with the mode.

## **13. Constant Value Imputation**



**Understand the Concept**

- Constant value imputation means replacing missing values with a fixed value.
- We can use values like `0`, `"Unknown"`, or `"Not Available"`.
- The value should be chosen based on the meaning of the column.

**When to use it**
- When a fixed value has a clear meaning.
- When missing values need to be kept as a separate category.

**When not to use it**
- When the chosen value can give a wrong meaning to the data.
- When it can affect the analysis or ML model.

**Advantages**
- Simple and easy to apply.
- Keeps all the rows in the dataset.

**Limitations**
- The chosen value may not represent the actual value.
- Can create an artificial pattern in the data.



**Demonstrate the Concept**

**Real-World Example:**
- Missing `Location` information can be marked as `Unknown`.

**AI/ML Usage:**
- Constant values can be used to handle missing numerical or categorical features.

```python
print("Missing Values Before:")
print(df["Location"].isnull().sum())
```

```python
# Implement the Concept
df["Location"] = df["Location"].fillna("Unknown")

print("\nMissing Values After:")
print(df["Furnishing"].isnull().sum())
```

**Explanation**

- fillna() replaces missing values with a fixed value.
- Here, missing Location values are replaced with `Unknown`.

## **14. Forward Fill**



**Understand the Concept**

- Forward fill replaces a missing value with the previous available value.
- It is mainly useful when the data follows an order or sequence.

**When to use it**
- When nearby previous values are suitable.
- For ordered or time-based data.

**When not to use it**
- When the previous value may not represent the missing value.
- For randomly ordered data.

**Advantages**
- Simple and easy to apply.
- Keeps the existing number of rows.

**Limitations**
- Can repeat the same value many times.
- May give incorrect results if values change frequently.



**Demonstrate the Concept**


**Real-World Example:**
- Parking_Spaces may have occasional missing entries.

**AI/ML Usage:**
- Forward fill can be used for ordered data before ML preprocessing.

```python
print("Missing Values Before:")
print(df["Parking_Spaces"].isnull().sum())
```

```python
# Implement the Concept
df["Parking_Spaces"] = df["Parking_Spaces"].ffill() # ffill() - forward fill

print("\nMissing Values After:")
print(df["Parking_Spaces"].isnull().sum())
```

**Explanation**

- ffill() fills missing values using the previous available value.
- It carries the previous value forward to the next missing row.

## **15. Backward Fill**



**Understand the Concept**

- Backward fill replaces a missing value with the next available value.
- It is useful when the next value is suitable for the missing record.

**When to use it**
- When the next available value is suitable.
- For ordered or time-based data.

**When not to use it**
- When the next value is unrelated to the missing value.
- For randomly ordered data.

**Advantages**
- Simple and easy to apply.
- Keeps the existing number of rows.

**Limitations**
- The next value may not always be suitable.
- It can repeat values across several rows.



**Demonstrate the Concept**

**Real-World Example:**
- A property record may have a missing value while the next record has a valid value.

**AI/ML Usage:**
- Backward fill can be used for suitable ordered data before ML preprocessing.

```python
print("Missing Values Before:")
print(df["Bedrooms"].isnull().sum())
```

```python
# Implement the Concept
df["Bedrooms"] = df["Bedrooms"].bfill()

print("\nMissing Values After:")
print(df["Bedrooms"].isnull().sum())
```

**Explanation**

- bfill() fills missing values using the next available value.
- It carries the next value backward to the missing row.

## **16. Interpolation**



**Understand the Concept**

- Interpolation estimates a missing value using nearby available values.
- It is mainly useful when the data follows an order or continuous pattern.

**When to use it**
- When values have a meaningful order.
- When nearby values can help estimate the missing value.

**When not to use it**
- When the data has no meaningful order.
- When nearby values are very different or irregular.

**Advantages**
- Uses nearby values to estimate missing data.
- Can give more realistic values than simple filling.

**Limitations**
- The estimated value may not be accurate.
- It may not work well with highly irregular data.



**Demonstrate the Concept**

**Real-World Example:**
- Property Distance_to_City_km may have occasional missing values between known measurements.

**AI/ML Usage:**
- Interpolation can be used to fill missing numerical values before ML preprocessing.

```python
print("Missing Values Before:")
print(df["Distance_to_City_km"].isnull().sum())
```

```python
# Implement the Concept
df["Distance_to_City_km"] = df["Distance_to_City_km"].interpolate()

print("\nMissing Values After:")
print(df["Distance_to_City_km"].isnull().sum())
```

**Explanation**

- interpolate() estimates missing values using nearby values.
- It fills missing numerical values without removing rows.

## **17. Group-Based Imputation**



**Understand the Concept**

- Group-based imputation fills missing values using information from a related group.
- It is useful when different groups have different typical values.
- The missing value is replaced using the mean, median, or mode of its group.

**When to use it**
- When the data has meaningful groups.
- When values differ between groups.

**When not to use it**
- When there are no useful groups.
- When some groups have too few records.

**Advantages**
- Uses group-specific information.
- Can be better than using one value for the whole dataset.

**Limitations**
- Results depend on the selected groups.
- Small groups may give unreliable values.



**Demonstrate the Concept**

**Example:**
- A house has a missing `Year_Built` value.
- We can use the median `Year_Built` of houses in the same `Location`.



**AI/ML Usage:**
- Group-based imputation can fill missing values while keeping differences between locations.

```python
print("Missing Values Before:")
print(df["Year_Built"].isnull().sum())
```

```python
# Implement the Concept
df["Year_Built"] = df["Year_Built"].fillna(df.groupby("Location")["Year_Built"].transform("median"))

print("\nMissing Values After:")
print(df["Year_Built"].isnull().sum())
```

**Explanation**

- groupby() creates groups based on Location.
- transform() calculates the median Year_Built for each location.
- fillna() fills missing values using the location-wise median.

## **18. KNN Imputation**



**Understand the Concept**

- KNN imputation fills missing values using similar rows in the dataset.
- It finds the nearest records based on other available features.

**When to use it**
- When related features can help estimate the missing value.
- When the dataset has enough similar records.

**When not to use it**
- When the dataset is very small or has many missing values.
- When the features are not suitable for measuring similarity.

**Advantages**
- Uses information from other features.
- Can give better results than simple imputation in some cases.

**Limitations**
- Can be slower for large datasets.
- Results can be affected by feature scales and outliers.



**Demonstrate the Concept**


**Real-World Example:**
- Houses with similar size and number of bedrooms may have similar prices.

**AI/ML Usage:**
- KNN imputation can fill missing numerical features before ML training.

```python
features = ["Area_sqft", "Bedrooms", "Bathrooms", "Owner_Count"]

print("Missing Values Before:")
print(df[features].isnull().sum())
```

```python
# Implement the Concept
from sklearn.impute import KNNImputer
imputer = KNNImputer(n_neighbors=5)
df[features] = imputer.fit_transform(df[features])
print("\nMissing Values After:")
print(df[features].isnull().sum())
```

**Explanation**

- KNNImputer finds similar records using the selected features.
- n_neighbors=5 uses the 5 nearest records.
- Missing Bathrooms values are estimated using similar houses.

## **19. Iterative Imputation**



**Understand the Concept**

- Iterative imputation fills missing values by using other columns in the dataset.
- It estimates the missing value several times and improves the estimate each time.
- It is useful when columns have relationships with each other.

**When to use it**
- When features are related to each other.
- When simple imputation may not give good results.

**When not to use it**
- When the dataset is very small.
- When there are too many missing values or weak relationships between features.

**Advantages**
- Uses relationships between multiple features.
- Can give better estimates than simple imputation.

**Limitations**
- Takes more time than simple imputation.
- Results depend on the quality of the other features.



**Demonstrate the Concept**

**Real-World Example:**
- Larger houses may generally have more bedrooms and bathrooms.

**AI/ML Usage:**
- Iterative imputation can estimate missing numerical features before ML training.

```python
# Implement the Concept
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

features = ["Area_sqft", "Bedrooms", "Bathrooms", "Owner_Count"]

imputer = IterativeImputer(random_state=42)
df[features] = imputer.fit_transform(df[features])
print("Missing Values:")
print(df[features].isnull().sum())
```

**Explanation**

- IterativeImputer estimates missing values using other features.
- It repeats the process to improve the estimates.
- This helps fill missing numerical values using relationships between columns.
