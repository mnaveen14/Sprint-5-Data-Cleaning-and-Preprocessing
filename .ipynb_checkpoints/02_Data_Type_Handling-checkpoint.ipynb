# **Notebook 2 – Data Type Handling**

```python
import pandas as pd
df = pd.read_csv("vehicle_records_dataset.csv")
df.info()
```

## **1. Identifying Data Types**



**Understand the Concept**

- Data types tell us what kind of data is stored in each column.
- We should check the data types before working with the dataset.



**Demonstrate the Concept**

**Example:**
- `Price` should be numeric.
- `Model` should be categorical or string.
- `Purchase_Date` should be a date/time value.

**Real-World Example:**
- Vehicle data can contain price, model, mileage, and purchase date.

**AI/ML Use Case:**
- ML models need data in the correct format before training.

**Why is it required?**
- Incorrect data types can cause errors and affect the analysis.

```python
print("Data Types:")
print(df.dtypes)
```

**Explanation**

- `dtypes` shows the data type of each column.
- This helps us find columns that have an incorrect data type.


## **2. Numerical Data**



**Understand the Concept**

- Numerical data contains numbers that we can use for calculations.
- It can be used to find things like average, minimum, maximum, and total.



**Demonstrate the Concept**

**Real-World Example:**
- Vehicle price and mileage are stored as numbers.

**AI/ML Use Case:**
- Numerical features can be used by ML models to make predictions.

**Why is it required?**
- Correct numerical data types allow us to perform calculations and use the values properly in ML models.

```python
print(df[["Year", "Price", "Mileage", "Engine_CC"]].dtypes)
```

**Explanation**

- `dtypes` shows the data type of each numerical column.
- Some numerical values may be stored as strings.
- We can identify them and convert them to numeric types.

## **3. Categorical Data**



**Understand the Concept**

- Categorical data contains values that represent groups or categories.
- These values are used to describe different types or groups.



**Demonstrate the Concept**

**Real-World Example:**
- Vehicle records use categories to describe the fuel type and transmission.


**AI/ML Use Case:**
- Categorical features can be converted into numbers before using them in an ML model.

**Why is it required?**
- Identifying categorical data helps us choose the right preprocessing and encoding method.

```python

print("Categorical Columns:")
print(df[["Model", "Fuel_Type", "Transmission", "City"]].dtypes)

print("\nUnique Values:")
print(df[["Model", "Fuel_Type", "Transmission", "City"]].nunique())
```

**Explanation**

- `dtypes` shows the data type of each categorical column.
- `nunique()` shows the number of unique categories.

## **4. Boolean Data**



**Understand the Concept**

- Boolean data has only two values: `True` or `False`.
- It is mainly used for yes/no or on/off information.



**Demonstrate the Concept**

**Real-World Example:**
- A vehicle record can show whether the vehicle is used or new.


**AI/ML Use Case:**
- Boolean values can be converted into numerical values when required by an ML model.

**Why is it required?**
- Identifying Boolean data helps us handle yes/no values correctly.

```python
print("Used Column Data Type:",df["Used"].dtype)

print("Unique Values:",df["Used"].unique())
```

**Explanation**

- `dtype` shows the data type of the `Used` column.
- `unique()` shows the different values in the column.

## **5. Date/Time Data**



**Understand the Concept**

- Date/Time data is used to store dates and time-related information.


**Demonstrate the Concept**


**Real-World Example:**
- Vehicle records can store the date when each vehicle was purchased.

**AI/ML Use Case:**
- Date information can be used to create features such as year, month, or day.

**Why is it required?**
- Correct date formats make it easier to perform date-based analysis and create useful features.

```python

print("Before Conversion:",df["Purchase_Date"].dtype)

df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"], errors="coerce", dayfirst=True)

print("After Conversion:",df["Purchase_Date"].dtype)
print("Invalid Dates:",df["Purchase_Date"].isnull().sum())
```

**Explanation**

- `to_datetime()` converts the values into date/time format.
- `errors="coerce"` changes invalid dates into missing values.
- `dayfirst=True` reads the date in day-month-year format.

## **6. String Data**



**Understand the Concept**

- String data contains text values such as names, models, or city names.



**Demonstrate the Concept**

**Real-World Example:**
- Vehicle records use text to store model names and city names.


**AI/ML Use Case:**
- String values can be converted into numerical values using encoding techniques.

**Why is it required?**
- Identifying string data helps us decide how to clean and process text-based columns.

```python
print("String Columns:")
print(df[["Vehicle_ID", "Model", "Fuel_Type", "City"]].dtypes)

print("\nString Values:")
print(df[["Vehicle_ID", "Model", "Fuel_Type", "City"]].head())
```

**Explanation**

- `dtypes` shows the data type of each string column.
- `head()` displays sample text values.

## **7. Converting Data Types**



**Understand the Concept**

- Data type conversion means changing a column from one data type to another.
- We convert data when the current type is not suitable for analysis or ML.


**Demonstrate the Concept**

**Real-World Example:**
- Vehicle data may have prices or years stored as text.

**AI/ML Use Case:**
- ML models require features to be in suitable data types for processing.

**Why is it required?**
- It helps avoid calculation errors and makes the data ready for further preprocessing.

```python
print("Before Conversion:")
print(df[["Year", "Price", "Mileage"]].dtypes)

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Mileage"] = pd.to_numeric(df["Mileage"], errors="coerce")

print("\nAfter Conversion:")
print(df[["Year", "Price", "Mileage"]].dtypes)
```

**Explanation**

- `to_numeric()` converts values into numeric data types.
- `errors="coerce"` changes invalid values into `NaN`.
- This helps convert numeric columns and identify invalid values.

## **8. `astype()`**



**Understand the Concept**

- `astype()` is used to change a column into a specific data type.
- It is useful when we know exactly which data type the column should have.



**Demonstrate the Concept**

**Real-World Example:**
- Vehicle year or price may be stored in the wrong data type.

**AI/ML Use Case:**
- `astype()` can prepare columns in the required format before ML processing.

**Why is it required?**
- It helps make sure the column has the expected data type.

```python
df["Owner_Count"] = pd.to_numeric(df["Owner_Count"], errors="coerce")

print("Before Conversion:")
print(df["Owner_Count"].dtype)

df["Owner_Count"] = df["Owner_Count"].astype("float")

print("\nAfter Conversion:")
print(df["Owner_Count"].dtype)
```

**Explanation**

- `astype()` changes the column to the specified data type.
- Here, `Owner_Count` is converted to `float`.

## **9. `to_numeric()`**



**Understand the Concept**

- `to_numeric()` is used to convert values into numeric data types.
- It is useful when numbers are stored as text.
- It can also help identify invalid values during conversion.



**Demonstrate the Concept**

**Real-World Example:**
- Vehicle prices may be stored as strings in a dataset.


**AI/ML Use Case:**
- Numeric features need to be in a suitable format before using them in ML models.

**Why is it required?**
- It allows us to perform calculations and work with numeric data correctly.

```python
print("Before Conversion:")
print(df["Price"].dtype)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

print("\nAfter Conversion:")
print(df["Price"].dtype)

print("\nInvalid Values:")
print(df["Price"].isnull().sum())
```

**Explanation**

- `to_numeric()` converts values into numeric format.
- `errors="coerce"` changes invalid values into `NaN`.
- This helps convert numeric columns and identify invalid values.

## **10. `to_datetime()`**



**Understand the Concept**

- `to_datetime()` is used to convert values into date and time format.
- It is useful when dates are stored as text.



**Demonstrate the Concept**

**Real-World Example:**
- Vehicle purchase dates may be stored as strings in a dataset.

**AI/ML Use Case:**
- Date values can be used to create features such as year, month, and day.

**Why is it required?**
- It helps us perform date-based analysis and handle dates correctly.

```python
print("Before Conversion:")
print(df["Purchase_Date"].dtype)

df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"],errors="coerce", dayfirst=True)
print("\nAfter Conversion:")
print(df["Purchase_Date"].dtype)

print("\nInvalid Dates:")
print(df["Purchase_Date"].isnull().sum())
```

**Explanation**

- `to_datetime()` converts values into date/time format.
- `errors="coerce"` changes invalid dates into `NaT`.
- `dayfirst=True` reads the date in day-month-year format.

## **11. Handling Invalid Conversions**



**Understand the Concept**

- Sometimes values cannot be converted because they contain text or invalid data.
- We need to identify these values before using the data.
- Invalid values can be converted to `NaN` so they can be handled later.



**Demonstrate the Concept**

**Real-World Example:**
- Vehicle price may contain values like `Unknown` or `Not Available`.


**AI/ML Use Case:**
- Handling invalid values prevents errors when preparing data for an ML model.

**Why is it required?**
- It helps identify and safely handle values that cannot be converted.

```python
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
print("Invalid Price Values:")
print(df[df["Price"].isnull()])
```

**Explanation**

- `to_numeric()` tries to convert the values into numbers.
- `errors="coerce"` changes invalid values into `NaN`.
- `isnull()` helps find the values that could not be converted.

**Understand the Concept**

**Demonstrate the Concept**

**AI/ML Use Case:**
