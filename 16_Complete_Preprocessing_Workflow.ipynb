## **5. Handle Missing Values**



**Understand the Concept**

- Missing values can prevent some ML algorithms from working correctly.
- Numerical features can be filled using the median.
- Categorical features can be filled using the most frequent value.

**Demonstrate the Concept**

**Example:**

- Fill missing numerical values with the median and categorical values with the most frequent value.

**AI/ML Usage:**

- Produces complete input features for the ML-ready dataset.

```python
# Handle missing values
for col in numeric_features:
    X[col] = X[col].fillna(X[col].median())

for col in categorical_features:
    X[col] = X[col].fillna(X[col].mode()[0])

print("Total Missing Values After Handling:", X.isnull().sum().sum())
```

## **9. Detect Outliers**



**Understand the Concept**

- Outliers are unusually high or low numerical values.
- They can affect statistical calculations and some ML algorithms.
- Detecting outliers helps decide whether they should be treated.

**Demonstrate the Concept**

**Example:**

- Use the IQR method to detect outliers in `Annual_Income`.

**AI/ML Usage:**

- Helps identify extreme values that may negatively affect model performance.

```python
# outliers using IQR

Q1 = X["Annual_Income"].quantile(0.25)
Q3 = X["Annual_Income"].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = X[(X["Annual_Income"] < lower_limit) | (X["Annual_Income"] > upper_limit)]

print("Number of Annual Income Outliers:", len(outliers))
```

## **10. Treat Outliers**



**Understand the Concept**

- Outliers can be treated by capping extreme values instead of removing rows.
- Capping keeps the observations while reducing the effect of extreme values.
- This is useful when extreme values may still contain meaningful information.

**Demonstrate the Concept**

**Example:**

- Cap `Annual_Income` using the IQR lower and upper limits.

**AI/ML Usage:**

- Helps reduce the influence of extreme numerical values while preserving the dataset.

```python
# Treat outliers using IQR capping

X["Annual_Income"] = X["Annual_Income"].clip(lower=lower_limit,upper=upper_limit)
print("Annual Income after outlier treatment:")
print(X["Annual_Income"].describe())
```

## **14. Perform Feature Selection**



**Understand the Concept**

- Feature selection removes features that provide little useful information.
- Removing unnecessary features can simplify the model.
- Feature selection should be performed using training data only to avoid data leakage.

**Demonstrate the Concept**

**Example:**

- Use `SelectKBest` to select the most useful numerical features.

**AI/ML Usage:**

- Helps reduce unnecessary features and can improve model efficiency.

```python
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, mutual_info_classif

# Select numerical features
X_numeric = X[numeric_features]

# Handle missing values before feature selection
imputer = SimpleImputer(strategy="median")
X_numeric_imputed = imputer.fit_transform(X_numeric)

# Feature selection
selector = SelectKBest(score_func=mutual_info_classif,k=min(5, len(numeric_features)))
X_selected = selector.fit_transform(X_numeric_imputed, y)
selected_features = X_numeric.columns[selector.get_support()]
print("Selected Features:")
print(selected_features.tolist())
print("\nSelected Feature Shape:", X_selected.shape)
```

## **15. Split Data**



**Understand the Concept**

- The dataset is divided into training and testing data.
- Training data is used to learn the model.
- Test data is kept unseen for final evaluation.
- Splitting should happen before fitting preprocessing steps to prevent data leakage.

**Demonstrate the Concept**

**Example:**

- Split the dataset into 80% training and 20% testing data.
- Use stratification to maintain the target class distribution.

**AI/ML Usage:**

- Provides unseen data for evaluating the final ML model.

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["Customer_ID", "Default"])
y = df["Default"]

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True))
print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True))
```

**Explanation**

## **16. Build Preprocessing Pipeline**



**Understand the Concept**

- A preprocessing pipeline combines multiple preprocessing steps into one workflow.
- Numerical features use imputation and scaling.
- Categorical features use imputation and encoding.
- The pipeline is fitted only on the training data.

**Demonstrate the Concept**

**Example:**

- Build a `ColumnTransformer` containing separate numerical and categorical pipelines.

**AI/ML Usage:**

- Creates a consistent and leakage-safe preprocessing workflow for ML models.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
numeric_features = X_train.select_dtypes(include="number").columns.tolist()
categorical_features = X_train.select_dtypes(include="object").columns.tolist()
```

```python
numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),("scaler", StandardScaler())])
categorical_pipeline = Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),("encoder", OneHotEncoder(handle_unknown="ignore"))])

preprocessor = ColumnTransformer([("num", numeric_pipeline, numeric_features),("cat", categorical_pipeline, categorical_features)])
```

```python
# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the same fitted pipeline
X_test_processed = preprocessor.transform(X_test)

print("Training Processed Shape:", X_train_processed.shape)
print("Testing Processed Shape:", X_test_processed.shape)
```
