# **Notebook 4 – Duplicate Data Handling**

```python
import pandas as pd
df = pd.read_csv("employee_dataset.csv")
df.info()
```

## **1. What are Duplicate Records?**



**Understand the Concept**

- Duplicate records are rows that contain the same or very similar information.
- They can occur because of repeated data entry, file merging, system errors, or data collection issues.



**Demonstrate the Concept**

**Example:**

- If the same employee record appears twice in a dataset, it is an **exact duplicate**.
- If two records have the same Name and Department but different Employee_IDs, they may be **partial duplicates**.

**AI/ML Usage:**

- Duplicate records can make some employees appear more frequently than they actually are.
- Removing or handling duplicates helps improve data quality before model training.



```python
# Implement the Concept
# Count duplicates
print("Total Rows:", len(df))
print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Show duplicates
df[df.duplicated(keep=False)].head(5)
```

**Explanation**

- duplicated() checks for repeated rows.
- sum() counts the duplicate rows.
- keep=False shows all duplicate records.

## **2. Exact Duplicates**



**Understand the Concept**

- Exact duplicates are rows where all column values are exactly the same.
- They usually happen because the same record is entered or imported more than once.



**Demonstrate the Concept**

**Real-World Example:**

- The same employee record is accidentally uploaded twice.

**AI/ML Usage:**

- Exact duplicates can give repeated importance to the same employee during analysis or model training.


```python
# Implement the Concept
# Exact duplicates
df[df.duplicated(keep=False)].sample(3)
```

**Explanation**

- duplicated() checks complete rows.
- keep=False shows all repeated rows.

## **3. Partial Duplicates**



**Understand the Concept**

- Partial duplicates are records where some important columns have the same values, but other columns are different.



**Demonstrate the Concept**

**Real-World Example:**

- An employee is entered twice with a different Employee_ID.
- Employee details are updated but the old record is still present.

**AI/ML Usage:**

- Partial duplicates can affect employee counts and analysis.
- They should be checked carefully before removing them.


```python
# Implement the Concept
# Partial duplicates
df[df.duplicated(subset=["Name", "Department"], keep=False)].sample(4)
```

**Explanation**

- subset checks duplicates using selected columns.
- keep=False shows all matching records.
- Other column values can still be different.

## **4. Identifying Duplicates**



**Understand the Concept**

- Duplicate identification means finding repeated records in a dataset.
- It helps us check how many duplicate records are present before removing them.



**Demonstrate the Concept**

**Real-World Example:**

- Find employees who were entered more than once.- Check repeated records before deleting them.

**AI/ML Usage:**

- Helps prevent repeated records from affecting data analysis.

**Why is it required?**

- To know whether duplicates exist.
- To inspect duplicate records before handling them.

```python
# Implement the Concept
# Count duplicates
print("Duplicate Rows:", df.duplicated().sum())
```

**Explanation**

- duplicated() finds repeated rows.

## **5. duplicated()**



**Understand the Concept**

- duplicated() is used to identify duplicate rows in a DataFrame.
- It returns True for rows that are duplicates and False for unique rows.



**Demonstrate the Concept**

**Example:**

- Use duplicated() to find repeated employee records.
- Use sum() to count the duplicate rows.

**AI/ML Usage:**

- Helps identify repeated data before model training.


```python
# Implement the Concept
# Find duplicates
duplicate_rows = df[df.duplicated()]
print("Duplicate Rows:", len(duplicate_rows))

# Show duplicates
duplicate_rows.sample(6)
```

**Explanation**

- duplicated() identifies repeated rows.
- It returns True for duplicate rows.

## **6. drop_duplicates()**



**Understand the Concept**

- drop_duplicates() is used to remove duplicate rows from a DataFrame.
- By default, it keeps the first occurrence and removes the repeated rows.



**Demonstrate the Concept**

**Real-World Example:**

- Remove an employee record that was accidentally entered twice.

**AI/ML Usage:**

- Helps prevent repeated records from affecting analysis and model training.

**Why is it required?**

- To remove unwanted duplicate records.
- To prepare clean data for further analysis.

```python
# Implement the Concept
# Rows before 
print("Before:", len(df))

# Remove duplicates
df_clean = df.drop_duplicates()

# Rows after 
print("After:", len(df_clean))
```

**Explanation**

- drop_duplicates() removes repeated rows.
- The first duplicate record is kept.

## **7. Duplicate Detection Based on Selected Columns**



**Understand the Concept**

- We can check duplicates using only specific columns.
- This is useful when all column values are not required to identify a duplicate.



**Demonstrate the Concept**

**Real-World Example:**

- Find employees with the same name working in the same department.

**AI/ML Usage:**

- Helps identify possible duplicate records using important features.
- Useful when exact row matching is not enough.

**Why is it required?**

- To find partial duplicates.
- To focus duplicate checking on important columns.

```python
# Implement the Concept
# Find selected-column duplicates
duplicates = df[df.duplicated(subset=["Name", "Department"], keep=False)]
duplicates[["Employee_ID", "Name", "Department", "Salary"]].head(10)
```

**Explanation**

- subset checks duplicates using selected columns.
- keep=False shows all matching records.
- Other column values can be different.

## **8. Handling Duplicate Records**



**Understand the Concept**

- Duplicate records should be checked before deciding to remove them.
- Some duplicates may be unwanted, while others may be valid records.



**Demonstrate the Concept**

**Real-World Example:**

- An employee record is accidentally entered twice.
- The duplicate copy can be removed after checking the details.

**AI/ML Usage:**

- Reduces repeated data before model training.

**Why is it required?**

- To remove unwanted duplicate records.

```python
# Implement the Concept
# Check duplicates
print("Before:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# Check again
print("After:", df.duplicated().sum())
```

**Explanation**

- First, duplicate rows are identified.
- drop_duplicates() removes exact duplicates.
- The final count shows whether duplicates remain.

## **9. Business Rules for Duplicate Removal**



**Understand the Concept**

Duplicate removal should be based on business context rather than blindly deleting repeated records. Some common rules:

- Keep the record with the most recent `Joining_Date` when it represents an updated employee record.
- Keep the record with the most complete employee information.
- Keep the record with the latest or valid `Performance_Score` when it represents a corrected entry.
- Keep multiple records when they represent valid employee events, such as transfers or role changes.



**Demonstrate the Concept**

**Example:**

- Two employee records may have the same Name and Department.
- Before deleting one, check the Employee_ID, Joining_Date, and other details.
- Remove a record only when the business rule confirms that it is an unwanted duplicate.
  
**AI/ML Usage:**

- Business rules help remove only unwanted duplicates.
- This prevents useful records from being lost before model training.


```python
# Implement the Concept
# Find possible duplicates
duplicates = df[df.duplicated(subset=["Name", "Department"], keep=False)]

# Sort by joining date
duplicates = duplicates.sort_values("Joining_Date")

# Show records
duplicates[["Employee_ID", "Name", "Department", "Joining_Date"]].head(10)
```

**Explanation**

- Name and Department help find possible duplicates.
- Joining_Date helps compare employee records.
- Business rules should decide which record to keep.

## **10. Risks of Blindly Deleting Duplicates**

**Understand the Concept**

Blindly deleting duplicates can be dangerous because not every repeated record is an error. Some duplicate-looking records may represent valid business information.

**Common risks:**

- Delete valid records that represent different employee events.
- Lose useful information when records contain different details.
- Incorrectly reduce employee counts or business activity.
- Affect AI/ML results by removing valid data.

**Demonstrate the Concept**

**Example:**

- Two employees may have the same Name and Department but different Employee_IDs.
- Deleting one record without checking the details may remove a valid employee record.
- Always check business rules before removing duplicates.
