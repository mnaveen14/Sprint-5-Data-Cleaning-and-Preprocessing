# **Notebook 8 – Feature Scaling**

```python
import pandas as pd
df = pd.read_csv("household_electricity_dataset.csv")
df.info()
```

## **1. Why Feature Scaling?**



**Understand the Concept**

- Feature scaling brings numerical features to a similar scale.
- It prevents features with larger values from dominating smaller ones.
- Scaling is important for algorithms that depend on distance or feature magnitude.

**Demonstrate the Concept**

**Example:**
- Compare Monthly_Income and Average_Temperature_C before scaling.

**AI/ML Usage:**
- Helps scale-sensitive models treat features more fairly.

```python
# Implement the Concept
# feature scaling before
print(df[["Monthly_Income","Average_Temperature_C"]].describe())
```

**Explanation**

- describe() shows the range and statistics of both features.
- Monthly_Income has much larger values than Average_Temperature_C.
- Feature scaling brings them to a comparable scale.

## **2. Scale-Sensitive Algorithms**



**Understand the Concept**

- Scale-sensitive algorithms are affected by the size and range of features.
- Features with larger values can have more influence on the model.

## **3. Scale-Insensitive Algorithms**



**Understand the Concept**

- Scale-insensitive algorithms are generally not affected by the scale of numerical features.
- They can work with features having different ranges without requiring scaling.
- Tree-based algorithms are common examples.

## **4. Normalization**



**Understand the Concept**

- Normalization scales values to a specific range, usually from 0 to 1.
- It is useful when features have different ranges.
- It is commonly used for distance-based and gradient-based algorithms.

**Demonstrate the Concept**

**Example:**
- Normalize Monthly_Electricity_Usage_kWh to a range between 0 and 1.

**AI/ML Usage:**
- Helps models compare features on a similar scale.

```python
# Implement the Concept
from sklearn.preprocessing import MinMaxScaler

# Create scaler
scaler = MinMaxScaler()

# Normalize feature
df["Usage_Normalized"] = scaler.fit_transform(df[["Monthly_Electricity_Usage_kWh"]])
df[["Monthly_Electricity_Usage_kWh","Usage_Normalized"]].head()
```

**Explanation**

- MinMaxScaler scales values between 0 and 1.
- fit_transform() calculates the range and applies scaling.
- The new column contains normalized values.

## **5. Standardization**



**Understand the Concept**

- Standardization changes values based on the mean and standard deviation.
- The resulting values usually have a mean of 0 and a standard deviation of 1.
- It is useful when features have different ranges and distributions.

**Demonstrate the Concept**

**Example:**
- Standardize Monthly_Income.

**AI/ML Usage:**
- Helps scale-sensitive models work better with features having different ranges.

```python
# Implement the Concept
# mean & std
mean = df["Monthly_Income"].mean()
std = df["Monthly_Income"].std()

# Apply standardization
df["Income_Standardized"] = (df["Monthly_Income"] - mean) / std
df[["Monthly_Income", "Income_Standardized"]].head()
```

**Explanation**

- Standardization uses the mean and standard deviation.
- The transformed values are centered around 0.
- It is useful for scale-sensitive algorithms.

## **6. Min-Max Scaling**

**Understand the Concept**

- Min-Max Scaling converts values to a fixed range, usually 0 to 1.
- It uses the minimum and maximum values of the feature.
- It is sensitive to extreme values.

**Demonstrate the Concept**

**Example:**
- Apply Min-Max Scaling to House_Area_sqft.

**AI/ML Usage:**
- Helps distance-based models compare features on a similar range.

```python
# Implement the Concept
from sklearn.preprocessing import MinMaxScaler
# MinMaxScaler 
scaler = MinMaxScaler()
df["Area_MinMax"] = scaler.fit_transform(df[["House_Area_sqft"]])
df[["House_Area_sqft", "Area_MinMax"]].head()
```

**Explanation**

- MinMaxScaler scales values between 0 and 1.
- fit_transform() learns the minimum and maximum values.
- Extreme values can affect the scaled results.

## **7. StandardScaler**



**Understand the Concept**

- StandardScaler standardizes features using the mean and standard deviation.
- The resulting feature is centered around 0.
- It is commonly used for scale-sensitive algorithms.

**Demonstrate the Concept**

**Example:**
- Apply StandardScaler to Monthly_Income.

**AI/ML Usage:**
- Helps algorithms such as KNN and SVM work with standardized features.

```python
# Implement the Concept
from sklearn.preprocessing import StandardScaler

# Standard Scaler 
scaler = StandardScaler()
df["Income_StandardScaler"] = scaler.fit_transform(df[["Monthly_Income"]])
df[[ "Monthly_Income","Income_StandardScaler"]].head()
```

**Explanation**

- StandardScaler calculates the mean and standard deviation.
- It transforms the feature to a standard scale.
- It is useful when feature ranges are very different.

## **8. MinMaxScaler**



**Understand the Concept**

- MinMaxScaler scales values to a selected range.
- The default range is 0 to 1.
- It is sensitive to outliers.

**Demonstrate the Concept**

**Example:**
- Apply MinMaxScaler to Monthly_Electricity_Usage_kWh.

**AI/ML Usage:**
- Helps distance-based models use features on a common range.

```python
# Implement the Concept
from sklearn.preprocessing import MinMaxScaler

# MinMax Scaler 
scaler = MinMaxScaler()
df["Usage_MinMaxScaler"] = scaler.fit_transform(df[["Monthly_Electricity_Usage_kWh"]])
df[["Monthly_Electricity_Usage_kWh", "Usage_MinMaxScaler"]].head()
```

**Explanation**

- MinMaxScaler uses the minimum and maximum values.
- Values are converted to a range between 0 and 1.
- Outliers can strongly affect the scaled range.

## **9. RobustScaler**



**Understand the Concept**

- RobustScaler scales features using the median and interquartile range.
- It is less affected by outliers than StandardScaler and MinMaxScaler.
- It is useful when numerical data contains extreme values.

**Demonstrate the Concept**

**Example:**
- Apply RobustScaler to Monthly_Electricity_Bill.

**AI/ML Usage:**
- Helps models handle features containing outliers more effectively.

```python
# Implement the Concept
from sklearn.preprocessing import RobustScaler
# Robust Scaler 
scaler = RobustScaler()
df["Bill_RobustScaler"] = scaler.fit_transform(df[["Monthly_Electricity_Bill"]])
df[["Monthly_Electricity_Bill","Bill_RobustScaler"]].head()
```

**Explanation**

- RobustScaler uses the median and interquartile range.
- It is less affected by extreme values.
- It is useful for data with outliers.

## **10. MaxAbsScaler**

**Understand the Concept**

- MaxAbsScaler scales values using the maximum absolute value.
- The values are generally placed between -1 and 1.
- It is useful when preserving zero values is important.

**Demonstrate the Concept**

**Example:**
- Apply MaxAbsScaler to Monthly_Income.

**AI/ML Usage:**
- Helps scale numerical features while preserving zero values.

```python
# Implement the Concept
from sklearn.preprocessing import MaxAbsScaler

# Max Abs Scaler 
scaler = MaxAbsScaler()
df["Income_MaxAbsScaler"] = scaler.fit_transform(df[["Monthly_Income"]])
df[["Monthly_Income","Income_MaxAbsScaler"]].head()
```

**Explanation**

- MaxAbsScaler divides values by the maximum absolute value.
- It keeps zero values unchanged.
- It is useful for data where zero has meaningful importance.
