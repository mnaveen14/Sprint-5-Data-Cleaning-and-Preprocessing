## **5. Data Consistency**



**Understand the Concept**

- Data consistency means keeping the same type of information in the same format.
- Inconsistent data can create problems during analysis and Machine Learning.
- We need to identify and fix values that are written differently.



**Demonstrate the Concept**

**Example:**

- These values should be changed to one consistent format.

**AI/ML Use Case:**

```python
print("Department Values Before Cleaning:")
print(df["Department"].value_counts(dropna=False))

df["Department"] = df["Department"].str.strip().str.title()

print("\nDepartment Values After Cleaning:")
print(df["Department"].value_counts(dropna=False))
```

**Explanation**

- `str.title()` makes the department names follow the same format.
- This helps keep categorical data consistent.

## **10. Data Leakage**



**Understand the Concept**

- Data leakage happens when information from outside the training data is used during model training.
- This can make the model perform very well during training but poorly on new data.
- We should avoid using future or test data while preparing the training data.



**Demonstrate the Concept**

**Example:**
- Using the test dataset to calculate the mean salary before splitting the data can cause leakage.

**Real-World Example:**
- Using information that was available only after an employee left the company to predict employee attrition.


**AI/ML Use Case:**
- Preprocessing steps such as imputation and scaling should be fitted only using the training data.

**Why is it required?**
- Preventing leakage helps us get a more realistic evaluation of the ML model.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
# Select numerical features
X = df[["Age", "Salary", "Experience_Years"]].copy()

# Handle missing values for this example
X = X.fillna(X.median())

# Split the data first
X_train, X_test = train_test_split(X, test_size=0.2, random_state=42)

# Fit the scaler only on training data
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training Data Shape:", X_train_scaled.shape)
print("Test Data Shape:", X_test_scaled.shape)
```

**Explanation**

- `train_test_split()` separates training and test data.
- `fit_transform()` learns the scaling values only from the training data.
- `transform()` applies the same scaling to the test data.

## **12. Validation Data**



**Understand the Concept**

- Validation data is used to check how well the model performs during development.
- It helps us compare different models and choose the better one.
- It is separate from the training data used to teach the model.



**Demonstrate the Concept**

**Example:**
- Employee data can be divided into training and validation data.
- The model learns from training data and is checked using validation data.

**AI/ML Use Case:**
- Validation data can be used for tuning model parameters and comparing model performance.

**Why is it required?**
- It helps evaluate the model during development without using the final test data.

```python
X = df[["Age", "Salary", "Experience_Years"]].copy()
X = X.fillna(X.median())

# First split: training and temporary data
X_train, X_temp = train_test_split(X, test_size=0.3, random_state=42)

# Second split: validation and test data
X_val, X_test = train_test_split(X_temp, test_size=0.5, random_state=42)

print("Training Data:", len(X_train))
print("Validation Data:", len(X_val))
print("Test Data:", len(X_test))
```

**Explanation**

## **14. Preprocessing Pipeline**



**Understand the Concept**

- A preprocessing pipeline is a sequence of steps used to prepare data for Machine Learning.
- It keeps all preprocessing steps organized and consistent.
- It helps apply the same steps correctly to new data.



**Demonstrate the Concept**

**Example:**
- Fill missing values first.
- Then scale numerical columns and encode categorical columns.

**Real-World Example:**
- Employee data may need missing value handling, scaling, and encoding before model training.


**AI/ML Use Case:**
- A pipeline can automatically prepare data before sending it to an ML model.

**Why is it required?**
- It reduces manual work and helps maintain a consistent preprocessing process.

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
numeric_features = ["Age", "Salary", "Experience_Years"]

numeric_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")),("scaler", StandardScaler())])

X = df[numeric_features]
X_processed = numeric_pipeline.fit_transform(X)
print("Original Shape:", X.shape)
print("Processed Shape:", X_processed.shape)
```

**Explanation**

- `Pipeline()` combines multiple preprocessing steps.
- `SimpleImputer()` handles missing values using the median.
- `StandardScaler()` scales the numerical features.
- `fit_transform()` learns the required values and applies the transformations.
- This makes the preprocessing steps easier to manage and reuse.
