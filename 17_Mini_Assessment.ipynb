## **1. What is Data Preprocessing?**

**Answer**

- Data preprocessing is the process of **cleaning, transforming, and preparing raw data** before using it for Machine Learning.
- It includes tasks such as **handling missing values, removing duplicates, treating outliers, encoding categorical variables, scaling numerical features, and selecting useful features**.
- Proper preprocessing improves **data quality and model performance** and helps create an **ML-ready dataset**.

## **2. Why is Data Preprocessing Required?**

**Answer**

- Data preprocessing is required to clean and prepare raw data before using it for Machine Learning.
- It helps handle missing values, duplicates, outliers, incorrect data, and categorical variables.
- Proper preprocessing improves data quality, model performance, and reliability.
- It also helps prevent problems such as data leakage and biased model results.
  

## **3. What are Missing Values?**

**Answer**

- Missing values are data points where no value is recorded for a particular feature.
- They can occur because of data collection errors, unavailable information, or incomplete records.
- Missing values can affect analysis and Machine Learning model performance.
- They should be identified and handled using an appropriate method.


## **4. Explain MCAR, MAR, and MNAR.**

**Answer**

- MCAR (Missing Completely At Random) means missing values occur randomly and are not related to any feature or observation.
- MAR (Missing At Random) means missing values depend on other observed features in the dataset.
- MNAR (Missing Not At Random) means missing values are related to the missing information itself.
- Understanding the type of missingness helps choose an appropriate handling method.



## **5. How Would You Handle a Feature Containing 40% Missing Values?**

**Answer**

- First, investigate why the feature has a high percentage of missing values.
- If the feature is not important, it may be removed from the dataset.
- If the feature is important, an appropriate imputation method can be considered.
- The decision should depend on the importance of the feature and the reason for the missing values.

## **6. What is the Difference Between Mean and Median Imputation?**

**Answer**

- Mean imputation replaces missing values with the average value of the feature.
- Median imputation replaces missing values with the middle value of the feature.
- Mean imputation can be affected by extreme values and outliers.
- Median imputation is more robust when the data contains outliers or is highly skewed.

## **7. What is Data Leakage?**

**Answer**

- Data leakage occurs when information that should not be available during model training is used by the model.
- It can happen when test data influences training or when future information is used as a feature.
- Data leakage can produce unrealistically high model performance.
- It should be prevented to ensure reliable evaluation on unseen data.

## **8. What is Target Leakage?**

**Answer**

- Target leakage occurs when a feature contains information that is directly or indirectly related to the target in a way that would not be available when making the prediction.
- It can make the model appear more accurate than it actually is.
- Future information related to the target should not be used as an input feature.
- Target leakage must be identified and removed before model training.

## **9. What is Train-Test Contamination?**

**Answer**

- Train-test contamination occurs when information from the test dataset influences the training process.
- This can happen when preprocessing is fitted on the complete dataset before splitting.
- It causes the test data to no longer represent completely unseen data.
- The test dataset should remain separate until final model evaluation.

## **10. What is Feature Leakage?**

**Answer**

- Feature leakage occurs when an input feature contains information that would not be available when the prediction is actually made.
- Such features can give the model information about the target that it should not have.
- Feature leakage can lead to misleadingly high model performance.
- Features should be checked carefully to ensure they are available at prediction time.

## **11. What is One-Hot Encoding?**

**Answer**

- One-Hot Encoding converts categorical values into separate binary columns.
- Each category is represented using 0 and 1.
- It is commonly used for nominal categorical features.
- It prevents the model from assuming an incorrect numerical order between categories.



## **12. What is the Difference Between Label Encoding and One-Hot Encoding?**

**Answer**

- Label Encoding assigns a numerical value to each category.
- One-Hot Encoding creates a separate binary column for each category.
- Label Encoding is suitable for ordinal categories where an order exists.
- One-Hot Encoding is generally suitable for nominal categories where no order exists.

## **13. What is Standardization?**

**Answer**

- Standardization transforms numerical features to have a mean close to 0 and a standard deviation close to 1.
- It is commonly performed using `StandardScaler`.
- It helps features with different ranges become comparable.
- It is useful for algorithms that are sensitive to feature scale.






## **14. What is Normalization?**

**Answer**

- Normalization transforms numerical values to a specific range, commonly between 0 and 1.
- It is commonly performed using `MinMaxScaler`.
- It is useful when features have different numerical ranges.
- It can be affected by extreme values and outliers.





## **15. When Would You Use RobustScaler?**

**Answer**

- `RobustScaler` is useful when numerical data contains outliers.
- It uses the median and interquartile range instead of the mean and standard deviation.
- It is less affected by extreme values.
- It can be preferred when the dataset contains significant outliers.





## **16. What is Class Imbalance?**

**Answer**

- Class imbalance occurs when the classes in the target variable are not represented equally.
- One class may contain significantly more observations than another.
- It can cause the model to favor the majority class.
- Techniques such as oversampling, undersampling, SMOTE, and class weights can help handle imbalance.





## **17. What is SMOTE?**

**Answer**

- SMOTE stands for Synthetic Minority Over-sampling Technique.
- It creates synthetic observations for the minority class.
- It helps increase minority-class representation in an imbalanced dataset.
- SMOTE should be applied only to the training data to avoid data leakage.




## **18. Why Can Accuracy Be Misleading for an Imbalanced Dataset?**

**Answer**

- Accuracy measures the overall percentage of correct predictions.
- In an imbalanced dataset, a model can achieve high accuracy by mainly predicting the majority class.
- This may result in poor detection of the minority class.
- Precision, recall, F1-score, and confusion matrix provide additional information about model performance.






## **19. Why Should Preprocessing Be Fitted Only on Training Data?**

**Answer**

- The test dataset should remain completely unseen during the training process.
- Fitting preprocessing on the complete dataset allows information from the test data to influence the training process.
- This can cause data leakage and produce overly optimistic results.
- Preprocessing should be fitted on training data and then used to transform both training and test data.




## **20. What is a Preprocessing Pipeline?**

**Answer**

- A preprocessing pipeline combines multiple preprocessing steps into a single workflow.
- It can include imputation, scaling, encoding, and other transformations.
- It ensures that the same preprocessing steps are applied consistently.
- It also helps reduce the risk of data leakage.



## **21. Why is ColumnTransformer Used?**

**Answer**

- `ColumnTransformer` allows different preprocessing methods to be applied to different groups of columns.
- Numerical features can be imputed and scaled.
- Categorical features can be imputed and encoded.
- It helps combine multiple preprocessing workflows into one pipeline.





## **22. What is the Difference Between fit(), transform(), and fit_transform()?**

**Answer**

- `fit()` learns the required parameters from the data.
- `transform()` applies the learned parameters to transform the data.
- `fit_transform()` performs both operations together.
- For test data, `transform()` should be used after fitting the preprocessing method on training data.





## **23. Why Should Duplicates Be Detected Before Model Training?**

**Answer**

- Duplicate records can cause the same observation to appear multiple times.
- They can give some observations more influence during model training.
- Duplicates can also affect statistical analysis and model evaluation.
- Removing unwanted duplicates helps create a cleaner ML-ready dataset.





## **24. What are Outliers?**

**Answer**

- Outliers are observations that are unusually different from the majority of the data.
- They can occur because of errors, unusual events, or genuine extreme observations.
- Outliers can affect statistical calculations and some Machine Learning algorithms.
- They should be investigated and treated when necessary.

## **25. Name Two Methods for Detecting Outliers.**

**Answer**

- Interquartile Range (IQR) method.
- Z-score method.
- Box plots can also be used to visually identify potential outliers.
- The appropriate method depends on the distribution and characteristics of the data.
